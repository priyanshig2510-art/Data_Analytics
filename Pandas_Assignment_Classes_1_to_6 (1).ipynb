# Q40. ⭐ Final Practical Challenge

Using the same 8-column DataFrame:

1. Sort employees by `Salary` from highest to lowest.
2. Find the average Salary of each Department.
3. Find the Department with the highest average Salary.
4. Find the average Score by City.
5. Create a pivot table of mean Salary by Department and City.
6. Create a crosstab of Department vs Gender.

**Expected Result:** A small analysis report containing all six results.

```python
# 1. Sort employees by Salary from highest to lowest
sorted_employees = df.sort_values(by='Salary', ascending=False)
print("--- 1. Sorted Employees by Salary ---")
print(sorted_employees.head(), "\n")

# 2. Find the average Salary of each Department
avg_salary_dept = df.groupby('Department')['Salary'].mean()
print("--- 2. Average Salary by Department ---")
print(avg_salary_dept, "\n")

# 3. Find the Department with the highest average Salary
highest_avg_salary_dept = avg_salary_dept.idxmax()
print("--- 3. Department with Highest Average Salary ---")
print(f"Department: {highest_avg_salary_dept}\n")

# 4. Find the average Score by City
# Note: Replace 'Score' with your actual column name if it differs (e.g., 'Performance_Score')
avg_score_city = df.groupby('City')['Score'].mean()
print("--- 4. Average Score by City ---")
print(avg_score_city, "\n")

# 5. Create a pivot table of mean Salary by Department and City
pivot_salary = df.pivot_table(values='Salary', index='Department', columns='City', aggfunc='mean')
print("--- 5. Pivot Table: Mean Salary by Dept and City ---")
print(pivot_salary, "\n")

# 6. Create a crosstab of Department vs Gender
crosstab_dept_gender = pd.crosstab(df['Department'], df['Gender'])
print("--- 6. Crosstab: Department vs Gender ---")
print(crosstab_dept_gender)
```
