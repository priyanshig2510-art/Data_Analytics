### Q1. Create two DataFrames `df1` and `df2` with columns `ID`, `Name`, `Marks`. df1 has Aman/Neha/Ravi and df2 has Priya/Karan/Simran.

**Expected Output: Two DataFrames with 3 rows each.**

```python
# Write your answer here
import pandas as pd

data1 = {'ID': [1, 2, 3], 'Name': ['Aman', 'Neha', 'Ravi'], 'Marks': [85, 92, 78]}
df1 = pd.DataFrame(data1)

data2 = {'ID': [4, 5, 6], 'Name': ['Priya', 'Karan', 'Simran'], 'Marks': [90, 88, 95]}
df2 = pd.DataFrame(data2)

print(df1)
print(df2)
```

### Q2. Use `pd.concat()` to combine `df1` and `df2` vertically.

**Expected Output: One DataFrame with 6 rows.**

```python
# Write your answer here
combine = pd.concat([df1, df2])
print(combine)
```

```python
# Write your answer here
ignoredf = pd.concat([df1, df2], ignore_index=True)
print(ignoredf)
```

### Q6. Create `students` with Student_ID, Name, Department and `marks` with Student_ID, Score. Use IDs 101–104 for students and 101,102,103,105 for marks.

**Expected Output: Both DataFrames are created.**

```python
# Write your answer here
student =pd.DataFrame({'Student_ID': [101, 102, 103, 104],
                       'Name': ['Aman', 'Neha', 'Ravi', 'Priya'],
                       'Department': ['CSE', 'ECE', 'MECH', 'CIVIL']})
```

### Q7. Perform an **inner merge** of students and marks using `Student_ID`.

**Expected Output: Only IDs 101, 102, 103 remain.**

```python
# Write your answer here
marks = pd.DataFrame({'Student_ID': [101, 102, 103, 105],
                      'Score': [85, 92, 78, 95]})
inner_merge = pd.merge(student, marks, on='Student_ID')
print(inner_merge)
```

### Q8. Perform a **left merge** using `Student_ID`.

**Expected Output: All students remain; ID 104 has missing Score.**

```python
# Write your answer here
left_merge = pd.merge(student, marks, on='Student_ID', how='left')
print(left_merge)
```

### Q9. Perform a **right merge** using `Student_ID`.

**Expected Output: All marks records remain; ID 105 has missing student information.**

```python
# Write your answer here
right_merge = pd.merge(student, marks, on='Student_ID', how='right')
print(right_merge)
```

### Q10. Perform an **outer merge** using `Student_ID`.

**Expected Output: IDs 101,102,103,104,105 are present.**

```python
# Write your answer here

outer_merge = pd.merge(student, marks, on='Student_ID', how='outer')
print(outer_merge)
```

### Q12. After the left merge, replace missing Score values with 0 using `fillna()`.

**Expected Output: Student 104 gets Score = 0.**

```python
# Write your answer here
left = pd.merge(student, marks, on='Student_ID', how='left')
left['Score'].fillna(0, inplace=True)
print(left)
```

### Q13. Create `employees` with `Emp_ID, Name` and `salary` with `Employee_ID, Salary`. Use IDs 1,2,3 in employees and 1,2,4 in salary. Merge using `left_on` and `right_on`.

**Expected Output: DataFrames merge using differently named key columns.**

```python
# Write your answer here

employees = pd.DataFrame({'Emp_ID': [1, 2, 3], 'Name': ['Aman', 'Neha', 'Ravi']})

salary = pd.DataFrame({'Employee_ID': [1, 2, 4], 'Salary': [50000, 60000, 70000]})

merge = pd.merge(employees, salary, left_on='Emp_ID', right_on='Employee_ID')
print(merge)
```

### Q14. Using the same employees and salary DataFrames, perform a left merge.

**Expected Output: All employees remain.**

```python
# Write your answer here

left = pd.merge(employees, salary, left_on='Emp_ID', right_on='Employee_ID', how='left')
print(left)
```

### Q15. Create `products` with Product_ID and Product, and `prices` with Product_ID and Price. Set Product_ID as the index in both.

### Q19. Create an `orders` DataFrame with Order_ID, Customer, Product and a `payments` DataFrame with Order_ID, Amount, Status. Use order IDs 1–4 for orders and 1,2,3,5 for payments.

**Expected Output: Two DataFrames with overlapping and non-overlapping Order_IDs.**

```python
# Write your answer here

orders = pd.DataFrame({'Order_ID': [1, 2, 3, 4],
                       'Customer': ['Aman', 'Neha', 'Ravi', 'Priya'],
                       'Product': ['A', 'B', 'C', 'D']})

payments = pd.DataFrame({
    'Order_ID': [1, 2, 3, 5],
    'Amount': [100, 200, 300, 400],
    'Status': ['Paid', 'Unpaid', 'Paid', 'Unpaid']
})

print(orders)
print(payments)
```

### Q20. For the orders/payments data: perform inner, left, and outer merges; check missing values; then identify orders that have no payment information.

**Expected Output: Unmatched Order_IDs can be identified from the merge.**

```python
# Write your answer here

inner_merge = pd.merge(orders, payments, on='Order_ID', how='inner')
left_merge = pd.merge(orders, payments, on='Order_ID', how='left')

print("Inner Merge:")
print(inner_merge)

print("\nLeft Merge:")
print(left_merge)
```
