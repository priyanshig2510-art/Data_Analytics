# Q1. Array Attributes

## Problem Statement
Create the following NumPy array:

```python
[10, 20, 30, 40, 50]
```

Print:
- Shape
- Number of Dimensions
- Size
- Data Type

## Sample Output
```python
Shape: (5,)
Dimensions: 1
Size: 5
dtype: int64
```


````python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print(f"Shape {arr.shape}")
print(f"dimensions {arr.ndim}")
print(f"Size {arr.size}")
print(f"dtype {arr.dtype}")



````

### Boolean Indexing (Practice)

**Q1.** Create a NumPy array from 1 to 20. Print only the elements greater than 10 using boolean indexing.



````python
arr1 = np.arange(1,21)
result = arr1[arr1 > 10]
print(result)
````


**Q2.** Create the array:
```python
arr = np.array([12, 5, 18, 7, 25, 30, 2, 15])
```
Print only the even numbers using boolean indexing.


````python
arr = np.array([12,5,18,7,25,30,2,151])
even = arr[arr % 2 == 0]
print(even)
````


**Q3.** Create a 4×4 random integer array (1–50). Print only the values greater than 25 using boolean indexing.


````python
arr = np.random.randint(1,51,size = (4 ,4))
print(arr)
values = arr [arr > 25]
print(values)
````
